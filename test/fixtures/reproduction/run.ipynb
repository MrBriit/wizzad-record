# Conductivity against road salt
Five readings, three sites.

```python
import csv, warnings
with open('data.csv') as f:
    rows = list(csv.DictReader(f))
print('rows:', len(rows))
print('sites:', len({r['site'] for r in rows}))
```

```python
warnings.warn('salt readings rounded to 0.5 g/L')
xs = [float(r['salt']) for r in rows]
ys = [float(r['uS']) for r in rows]
n = len(xs)
mx, my = sum(xs) / n, sum(ys) / n
slope = sum((x - mx) * (y - my) for x, y in zip(xs, ys)) / sum((x - mx) ** 2 for x in xs)
round(slope, 2)
```

```python
from IPython.display import display
by_site = {}
for r in rows:
    by_site.setdefault(r['site'], []).append(int(r['uS']))
print('highest:', max(ys))
display({k: max(v) for k, v in sorted(by_site.items())})
```

```python
intercept = my - slope * mx
{'slope': round(slope, 2), 'intercept': round(intercept, 1)}
```
